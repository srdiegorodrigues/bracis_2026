```python
# Excluir registros (latitude e longitude) fora dos limites territoriais do Brasil

def excluir_registros_fora_brasil(df_clean):
    lat_min, lat_max = -33.750, 5.272
    long_min, long_max = -73.990, -34.793

    # Verificar se a latitude ou longitude está fora dos limites do Brasil e excluir o registro
    df_clean = df_clean[~((df_clean['latitude'] < lat_min) | (df_clean['latitude'] > lat_max) |
              (df_clean['longitude'] < long_min) | (df_clean['longitude'] > long_max))]

    return df_clean

# Aplicar a função ao DataFrame
df_clean = excluir_registros_fora_brasil(df_clean)
```

```python
# Função para fazer o tratamento do atributo tracado_via que possui mais  de 90 
# possibilidades de resposta

def extrair_valores_unicos(df, coluna):
    valores_unicos = set()

    for entrada in df[coluna]:
        valores = entrada.split(';')
        for valor in valores:
            valores_unicos.add(valor.strip())

    return list(valores_unicos)

def adicionar_colunas_valores_unicos(df, coluna):
    # Extrair valores únicos
    valores_unicos = extrair_valores_unicos(df, coluna)

    # Adicionar novas colunas ao DataFrame e inicializar com zero
    for valor in valores_unicos:
        df[valor] = 0

    # Atualizar as colunas de acordo com os valores em 'tracado_via'
    for index, entrada in df.iterrows():
        valores = entrada[coluna].split(';')
        for valor in valores:
            valor = valor.strip()
            if valor in df.columns:
                df.at[index, valor] = 1

    return df

df_clean = adicionar_colunas_valores_unicos(df_clean, 'tracado_via')
```

```python
#Verifica se o dia do acidente foi em um feriado ou não

import holidays

def add_feriado_column(df, country='BR'):
    # Inicializa os feriados nacionais e estaduais
    national_holidays = holidays.CountryHoliday(country, years=range(df['data_inversa'].dt.year.min(), df['data_inversa'].dt.year.max() + 1))
    state_holidays = {state: holidays.CountryHoliday(country, subdiv=state, years=range(df['data_inversa'].dt.year.min(), df['data_inversa'].dt.year.max() + 1)) for state in df['uf'].unique()}

    # Função para calcular o valor do atributo 'feriado'
    def calculate_feriado(row):
        date = pd.Timestamp(row['data_inversa'])
        state = row['uf']

        # Verifica se a data é um feriado nacional
        if date in national_holidays:
            return 0

        # Verifica se a data é um feriado estadual
        elif date in state_holidays[state]:
            return 0

        # Se não for feriado, calcula a distância para o feriado mais próximo
        all_holidays = list(national_holidays) + list(state_holidays[state])
        all_holidays = [pd.Timestamp(h) for h in all_holidays]

        # Calcula a distância para o feriado mais próximo (antes ou depois)
        distances = [(date - h).days for h in all_holidays]

        # Encontra a menor distância com o sinal ajustado
        abs_distances = [abs(d) for d in distances]
        min_abs_distance = min(abs_distances)

        # Verifica qual valor (positivo ou negativo) é menor
        closest_distance = min((d for d in distances if abs(d) == min_abs_distance), key=lambda x: x)

        return closest_distance

    # Aplica a função a cada linha do DataFrame
    df['feriado'] = df.apply(calculate_feriado, axis=1)

    return df

# Converte a coluna 'data_inversa' para o formato datetime
df_clean['data_inversa'] = pd.to_datetime(df_clean['data_inversa'])

# Adiciona a coluna 'feriado'
df_clean = add_feriado_column(df_clean, 'BR')
```

```python
# Criação do atributo turno
# em substituição do atributo fase_dia 

def atribuir_turno(df):
    # Define a função para determinar o turno com base no horário
    def determinar_turno(horario):
        if pd.isnull(horario):
            return None
        elif '00:00:00' <= horario <= '05:59:00':
            return 'madrugada'
        elif '06:00:00' <= horario <= '11:59:00':
            return 'manhã'
        elif '12:00:00' <= horario <= '17:59:00':
            return 'tarde'
        elif '18:00:00' <= horario <= '23:59:00':
            return 'noite'
        elif horario >= '24:00:00':
            return 'madrugada'
        else:
            print(horario)

    # Aplica a função para criar a coluna "turno"
    df['turno'] = df['horario'].apply(determinar_turno)
    return df

df_clean = atribuir_turno(df_clean)
```

```python
#arredondar o valor de km para cima ou para baixo, conforme o valor após o ponto
def arredondar_km(valor):
    valor_float = float(valor)  # Converter para float
    inteiro = int(valor_float)  # parte inteira do valor
    decimal = valor_float - inteiro  # parte decimal do valor

    # Se o valor decimal for maior ou igual a 0.5, arredonda para cima, caso contrário, arredonda para baixo
    if decimal >= 0.5:
        return inteiro + 1
    else:
        return inteiro
```

```python
#Função para determinar a velocidade máxima e mínima da via para o veículo
#envolvido no acidente

def determinar_velocidade(tipo_pista, uso_solo, tipo_veiculo):
    # Configurações de velocidades máximas e mínimas conforme CTB
    velocidades = {
        'urbano': {
            'Simples': {'Automóvel': (40, 60), 'Motocicleta': (40, 60), 'Caminhonete': (40, 60),
                        'Camioneta': (40, 60), 'Ônibus': (40, 50), 'Micro-ônibus': (40, 50),
                        'Caminhão': (40, 50), 'Utilitário': (40, 50)},
            'Dupla': {'Automóvel': (40, 80), 'Motocicleta': (40, 80), 'Caminhonete': (40, 80),
                      'Camioneta': (40, 80), 'Ônibus': (40, 60), 'Micro-ônibus': (40, 60),
                      'Caminhão': (40, 60), 'Utilitário': (40, 60)},
            'Múltipla': {'Automóvel': (40, 80), 'Motocicleta': (40, 80), 'Caminhonete': (40, 80),
                         'Camioneta': (40, 80), 'Ônibus': (40, 60), 'Micro-ônibus': (40, 60),
                         'Caminhão': (40, 60), 'Utilitário': (40, 60)}
        },
        'rural': {
            'Simples': {'Automóvel': (60, 100), 'Motocicleta': (60, 100), 'Caminhonete': (60, 100),
                        'Camioneta': (60, 100), 'Ônibus': (50, 90), 'Micro-ônibus': (50, 90),
                        'Caminhão': (50, 80), 'Utilitário': (50, 80)},
            'Dupla': {'Automóvel': (60, 110), 'Motocicleta': (60, 110), 'Caminhonete': (60, 110),
                      'Camioneta': (60, 110), 'Ônibus': (50, 90), 'Micro-ônibus': (50, 90),
                      'Caminhão': (50, 90), 'Utilitário': (50, 90)},
            'Múltipla': {'Automóvel': (60, 110), 'Motocicleta': (60, 110), 'Caminhonete': (60, 110),
                         'Camioneta': (60, 110), 'Ônibus': (50, 90), 'Micro-ônibus': (50, 90),
                         'Caminhão': (50, 90), 'Utilitário': (50, 90)}
        }
    }

    # Determinar se é área urbana ou rural
    area = 'urbano' if uso_solo == 'sim' else 'rural'

    # Se o tipo de veículo for NaN, consideramos como Automóvel
    if pd.isna(tipo_veiculo):
        tipo_veiculo = 'Automóvel'

    # Definimos velocidades padrão caso o veículo não seja categorizado
    vel_padrao = (30, 40)

    # Buscar as velocidades no dicionário
    vel_min, vel_max = velocidades.get(area, {}).get(tipo_pista, {}).get(tipo_veiculo, vel_padrao)

    return vel_max, vel_min

def aplicar_velocidades(df):
    df['vel_max'], df['vel_min'] = zip(*df.apply(
        lambda row: determinar_velocidade(row['tipo_pista'], row['uso_solo'], row['tipo_veiculo']), axis=1))
    return df

# Exemplo de uso:
df_clean = aplicar_velocidades(df_clean)
```

```python
# Converter latitude e longitude para float
def converter_para_float(df, coluna):
    for indice, valor in df[coluna].items():
        if isinstance(valor, str):  # Verifica se o valor é do tipo string
            try:
                valor = valor.replace(',', '.')  # Substitui vírgula por ponto
                df.at[indice, coluna] = float(valor)  # Converte para float
            except ValueError:  # Captura exceção se a conversão falhar
                print(f"Erro ao converter valor '{valor}' para float no índice {indice} na coluna {coluna}")

    # Garante que toda a coluna seja convertida para float
    df[coluna] = pd.to_numeric(df[coluna], errors='coerce')
    return df

# Aplicar a função de conversão nas colunas de latitude e longitude
df_radares = converter_para_float(df_radares, 'latitude')
df_radares = converter_para_float(df_radares, 'longitude')
```

```python
# Determina a distância em metros do ponto de ocorrência do sinistro 
# até o radar mais próximo

import geopandas as gpd

def encontrar_radar_mais_proximo(df_acidente, df_radares):

    # Converte para GeoDataFrame com CRS geográfico
    gdf_acidente = gpd.GeoDataFrame(
        df_acidente,
        geometry=gpd.points_from_xy(df_acidente["longitude"], df_acidente["latitude"]),
        crs="EPSG:4326"
    )

    gdf_radares = gpd.GeoDataFrame(
        df_radares,
        geometry=gpd.points_from_xy(df_radares["longitude"], df_radares["latitude"]),
        crs="EPSG:4326"
    )

    # Reprojeta para sistema métrico (UTM Brasil - ajuste se necessário)
    gdf_acidente = gdf_acidente.to_crs(epsg=31983)
    gdf_radares = gdf_radares.to_crs(epsg=31983)

    # Inicializa colunas
    gdf_acidente["latitude_radar"] = np.nan
    gdf_acidente["longitude_radar"] = np.nan
    gdf_acidente["distancia_radar_m"] = np.nan

    for i, row in gdf_acidente.iterrows():

        radares_filtrados = gdf_radares[gdf_radares["rodovia"] == row["br"]]

        if radares_filtrados.empty:
            radares_filtrados = gdf_radares

        # Distância agora está em METROS
        distancias = row.geometry.distance(radares_filtrados.geometry)

        idx_min = distancias.idxmin()
        radar_mais_proximo = radares_filtrados.loc[idx_min]

        gdf_acidente.loc[i, "distancia_radar_m"] = distancias.loc[idx_min]

        # Converter coordenadas do radar de volta para lat/long
        radar_geo = radar_mais_proximo.geometry
        radar_geo_wgs84 = gpd.GeoSeries([radar_geo], crs=gdf_acidente.crs).to_crs(epsg=4326)

        gdf_acidente.loc[i, "latitude_radar"] = radar_geo_wgs84.y.values[0]
        gdf_acidente.loc[i, "longitude_radar"] = radar_geo_wgs84.x.values[0]

    return gdf_acidente


df_clean = encontrar_radar_mais_proximo(df_clean, df_radares)
```

```python
def salvar_value_counts_completo(df, caminho_arquivo, incluir_nulos=False):
    """
    Percorre todas as colunas do DataFrame,
    aplica value_counts() e salva o resultado completo
    em um arquivo texto.
    """

    with open(caminho_arquivo, 'w', encoding='utf-8') as arquivo:

        colunas = df.columns.difference(['id', 'pesid', 'data_inversa', 'dia', 
                                         'geometry', 'latitude_radar', 'longitude_radar', 
                                         'distancia_radar_m', 'geometry''distancia_radar_m',
                                         'horario','id_veiculo', 'latitude','longitude',
                                         'km','marca','municipio'])

        for coluna in colunas:

            arquivo.write(f"{coluna}\n")

            # Calcula as frequências
            contagem = df[coluna].value_counts(dropna=not incluir_nulos)

            # Converte para string preservando o formato do pandas
            arquivo.write(contagem.to_string())
            arquivo.write("\n\n")


caminho = '../database/valores_atributos/valores_atributos.txt'
salvar_atributos = salvar_value_counts_completo(df_clean,caminho, True)
```
