```python
#Importação das bibliotecas
import pandas as pd
import numpy as np
import io
```

```python
url = "../../database/completo/df_atributos_considerados.csv"

# carrega dados
df = pd.read_csv( url, sep=",", encoding="UTF-8")
```

```python
# tratamento das condições meteorológicas

# Dicionário de mapeamento
mapa_condicao = {
    'Céu Claro': 'Sol',
    'Sol': 'Sol',
    'Nublado': 'Nublado',
    'Chuva': 'Chuva',
    'Garoa/Chuvisco': 'Chuva',
    'Nevoeiro/Neblina': 'Neblina',
    'Vento': 'Neblina',
    'Granizo': 'Chuva',
    'Neve': 'Chuva',
    'Ignorado': 'Sol'  # ou 'Missing'
}

# Aplicando transformação
df['condicao_metereologica'] = df['condicao_metereologica'].map(mapa_condicao)
```

```python
# mapeamento tipo_pista

mapa_tipo_pista = {
  'Simples':2,
  'Dupla':1,
  'Múltipla': 0
}

# Aplicando transformação
df['tipo_pista'] = df['tipo_pista'].map(mapa_tipo_pista)
```

```python
# mapeamento turno

mapa_turno = {
  'madrugada': 3,
  'manhã':2,
  'tarde': 1,
  'noite': 0
}

# Aplicando transformação
df['turno'] = df['turno'].map(mapa_turno)
```

```python
# tratamento das condições meteorológicas

# Dicionário de mapeamento
mapa_sentido_via = {
    'Crescente': 0,
    'Decrescente': 1
}

# Aplicando transformação
df['sentido_via'] = df['sentido_via'].map(mapa_sentido_via)
```

```python
# tratamento das condições meteorológicas

# Dicionário de mapeamento
mapa_uso_solo = {
    'Não': 0,
    'Sim': 1
}

# Aplicando transformação
df['uso_solo'] = df['uso_solo'].map(mapa_uso_solo)
```

```python
# Dicionário de mapeamento -> dia semana
mapa_dia_semana = {
    'segunda-feira': 0,
    'terça-feira': 1,
    'quarta-feira': 2,
    'quinta-feira': 3,
    'sexta-feira': 4,
    'sábado': 5,
    'domingo': 6,
}

# Aplicando transformação
df['dia_semana'] = df['dia_semana'].map(mapa_dia_semana)
```

```python
# Dicionário de mapeamento ->condicao metereologica
mapa_condi_meteorologica= {
    'Sol': 0,
    'Nublado': 1,
    'Chuva': 2,
    'Neblina': 3,
}

# Aplicando transformação
df['condicao_metereologica'] = df['condicao_metereologica'].map(mapa_condi_meteorologica)
```

```python
df["uf_br"] = df["uf"].astype(str) + "-" + df["br"].astype(int).astype(str)
df = df.drop(columns=["uf", "br"])

codigos, categorias = pd.factorize(df["uf_br"], sort=True)
df["uf_br"] = codigos  # inteiro ordinal temporário
```

```python
# Tratamento do nome dos atributos

import unidecode

# Supondo que df_agrupados já esteja carregado
# Exemplo de DataFrame

# Função para modificar os nomes das colunas
def formatar_nome(nome):
    nome = unidecode.unidecode(nome)  # Remove acentuação
    nome = nome.lower()  # Converte para minúsculas
    nome = nome.replace(' ', '_')  # Substitui espaços por underscore
    nome = nome.replace('-', '_')  # Substitui espaços por underscore
    return nome

# Aplica a formatação nas colunas do DataFrame
df.columns = [formatar_nome(col) for col in df.columns]

# Exibe as novas colunas
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

colunas = ['uf_br','km', 'dia_semana','vel_max', 'distancia_radar_m', 'total_ocorrencia',
           'feriado','mes', 'turno','condicao_metereologica','tipo_pista']

df[colunas] = scaler.fit_transform(df[colunas])
```

```python
df = df.dropna()
```

```python
# Salvando como CSV, sem o índice (index=False)
df.to_csv('../../database/completo/df_completo_01_uf_br.csv', index=False, encoding='utf-8')
```
