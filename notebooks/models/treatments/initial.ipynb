```python
#Importação das bibliotecas
import pandas as pd
import numpy as np
import io
```

```python
url = "../../database/"

# carrega dados
df_2017 = pd.read_csv(io.StringIO(open(f'{url}acidentes2017_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2018 = pd.read_csv(io.StringIO(open(f'{url}acidentes2018_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2019 = pd.read_csv(io.StringIO(open(f'{url}acidentes2019_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2020 = pd.read_csv(io.StringIO(open(f'{url}acidentes2020_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2021 = pd.read_csv(io.StringIO(open(f'{url}acidentes2021_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2022 = pd.read_csv(io.StringIO(open(f'{url}acidentes2022_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2023 = pd.read_csv(io.StringIO(open(f'{url}acidentes2023_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2024 = pd.read_csv(io.StringIO(open(f'{url}acidentes2024_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')
df_2025 = pd.read_csv(io.StringIO(open(f'{url}acidentes2025_todas_causas_tipos.csv', 'r', encoding='ISO-8859-1').read()), sep=';', on_bad_lines='skip')

# combina todos os registros
list_sinistros_years = [df_2017, df_2018, df_2019, df_2020, df_2021, df_2022, df_2023, df_2024, df_2025]
df = pd.concat(list_sinistros_years, ignore_index=True)
```

```python
# Remove os pesid repetidos na base de dados

def remove_duplicate_pesids(df):
    # # Verificar se a coluna 'pesid' existe no dataset
    # if 'pesid' not in df.columns:
    #     raise ValueError("O dataset não contém uma coluna 'pesid'.")

    # Remover duplicados mantendo apenas o primeiro registro de cada 'pesid'
    df_clean = df.drop_duplicates(subset='id', keep='first')

    return df_clean
df_2017 = remove_duplicate_pesids(df_2017)
df_2018 = remove_duplicate_pesids(df_2018)
df_2019 = remove_duplicate_pesids(df_2019)
df_2020 = remove_duplicate_pesids(df_2020)
df_2021 = remove_duplicate_pesids(df_2021)
df_2022 = remove_duplicate_pesids(df_2022)
df_2023 = remove_duplicate_pesids(df_2023)
df_2024 = remove_duplicate_pesids(df_2024)
df_2025 = remove_duplicate_pesids(df_2025)
```
