```python
#Importação das bibliotecas
import pandas as pd
import numpy as np
import io
```

```python
url = "../../database/completo/df_treatment_part01.csv"

df = pd.read_csv( url, sep=",", encoding="UTF-8")
```

```python
consideradas = ['dia_semana', 'uf','br', 'km', 'sentido_via', 'pista_molhada','visibilidade_ruim',
    'tipo_pista',  'uso_solo', 'intersecao_de_vias', 'reta', 'declive','condicao_ignorada',
    'feriado', 'turno', 'ano', 'mes', 'vel_max', 'distancia_radar_m', 'frequencia', 'grave']

id_desconsiderados = ['id', 'pesid', 'id_veiculo']

originaram_outros = ['data_inversa', 'fase_dia','horario', 'tracado_via', 'latitude','uf','br','condicao_metereologica',
    'longitude',]

muita_granularidade = ['municipio', 'delegacia', 'uop','da',] 

oriundo_sinistro = ['causa_principal', 'causa_acidente', 'ordem_tipo_acidente', 'tipo_acidente',
'tipo_envolvido', 'estado_fisico', 'idade', 'sexo', 'ilesos', 'feridos_leves', 'feridos_graves',
'mortos', 'classificacao_acidente'] 

baixa_ocorrencia = ['em_obras', 'rotatoria','retorno_regulamentado', 'viaduto', 
'desvio_temporario', 'ponte', 'desvio_temporario']

multicolinearidade = ['aclive', 'tunel', 'curva', 'latitude_radar', 'longitude_radar', 'geometry','vel_min'] 

dados_veiculo = ['tipo_veiculo', 'marca','ano_fabricacao_veiculo', 'regional']
```

```python
df_new = df[consideradas].copy()
```

```python
df_new.visibilidade_ruim.value_counts()
```

```python
df_new["km"] = (
    df_new["km"]
    .str.replace(",", ".", regex=False)
)

df_new["km"] = pd.to_numeric(df_new["km"], errors="coerce")

df_new["km"] = df_new["km"].round().astype("Int64")


df_new["br"] = pd.to_numeric(df_new["br"], errors="coerce")
```

```python
# df['distancia_radar_m'] = df['distancia_radar_m'].apply(lambda x: f"{x:.2f}")
df_new['distancia_radar_m'] = df_new['distancia_radar_m'].round(2)
```
